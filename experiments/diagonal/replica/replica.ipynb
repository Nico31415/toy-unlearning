```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 5))

# ---------- empirical summary ----------
emp_summary = (
    empirical_df
    .groupby(['alpha', 'c_pt'])['final_param_mse']
    .agg(['mean', 'sem'])
    .reset_index()
)

# color cycle
colors = plt.rcParams['axes.prop_cycle'].by_key()['color']

for i, c_pt in enumerate(sorted(emp_summary['c_pt'].unique())):
    color = colors[i % len(colors)]

    # empirical: points + SEM
    df_emp = emp_summary[emp_summary['c_pt'] == c_pt]
    plt.errorbar(
        df_emp['alpha'],
        df_emp['mean'],
        yerr=df_emp['sem'],
        fmt='o',
        capsize=3,
        linestyle='none',
        color=color,
        label=f'Empirical (c={c_pt})'
    )

    # theory c = 1
    df_th = df_replica_st_c1[df_replica_st_c1['c'] == c_pt]
    if not df_th.empty:
        plt.plot(
            df_th['alpha'],
            df_th['mse_best'],
            linestyle='--',
            linewidth=2,
            color=color,
            label=f'Theory c=1 (c={c_pt})'
        )

    # theory c = 1e-3
    df_th2 = df_replica_st_c1e3[df_replica_st_c1e3['c'] == c_pt]
    if not df_th2.empty:
        plt.plot(
            df_th2['alpha'],
            df_th2['mse_best'],
            linestyle='--',
            linewidth=2,
            color=color,
            label=f'Theory c=1e-3 (c={c_pt})'
        )

plt.xlabel('alpha')
plt.ylabel('MSE')
plt.legend()
plt.tight_layout()
plt.show()
```
